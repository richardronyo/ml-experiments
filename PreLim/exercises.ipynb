9. This exercise involves the Auto data set studied in the lab. Make sure
that the missing values have been removed from the data.

```python
auto = pd.read_csv("Auto.csv", na_values = '?')
auto = auto.dropna()
```

```python
#a
auto.dtypes
quantitative_cols = auto.select_dtypes(include=['float64', 'int64']).columns
qualitative_cols = auto.select_dtypes(include=['str']).columns

print("Qualitative Predictors: ", qualitative_cols)
print("Quantitative Predictors: ", quantitative_cols)
```

```python
#b
for col in quantitative_cols:
    print(f"Range of {col}: {auto[col].min()} to {auto[col].max()}")
```

```python
#c
print(f"{'Predictor':<20} {'Mean':>10} {'Std Dev':>10}")
print("-" * 42)
for col in quantitative_cols:
    mean = auto[col].mean()
    std = auto[col].std()
    print(f"{col:<20} {mean:>10.2f} {std:>10.2f}")
```

```python
#d
altered_auto = auto.drop(auto.iloc[10:85].index)

for col in quantitative_cols:
    print(f"Range of {col}: {altered_auto[col].min()} to {altered_auto[col].max()}")

print(f"\n{'Predictor':<20}\t{'Mean':<10}\t{'Standard Deviation':<10}")
print("-" * 80)
for col in quantitative_cols:
    mean = altered_auto[col].mean()
    std = altered_auto[col].std()
    print(f"{col:<20}\t{mean:<10}\t{std:<10}")
```

```python
#e
correlation_matrix = auto.corr(numeric_only = True)
fig, ax = plt.subplots(figsize = (30, 10))
ax.imshow(correlation_matrix, cmap = 'coolwarm')
ax.set_xticks(range(len(correlation_matrix.columns)))
ax.set_yticks(range(len(correlation_matrix.columns)))
ax.set_xticklabels(correlation_matrix.columns)
ax.set_yticklabels(correlation_matrix.columns)
plt.colorbar(ax.imshow(correlation_matrix, cmap = 'coolwarm'))
```

```python
correlated_columns = ['cylinders', 'displacement', 'horsepower', 'weight']
pairs = [(col1, col2) for i, col1 in enumerate(correlated_columns) 
                      for col2 in correlated_columns[i+1:]]

nplots = len(pairs)  # 6 pairs for 4 columns
fig, ax = plt.subplots(nrows=2, ncols=3, figsize=(15, 10))

for idx, (col1, col2) in enumerate(pairs):
    row, col = divmod(idx, 3)
    ax[row, col].scatter(auto[col1], auto[col2], alpha=0.5)
    ax[row, col].set_xlabel(col1)
    ax[row, col].set_ylabel(col2)

plt.tight_layout()
plt.show()
```
