Preliminary MatPlotLib

```python
fig, ax = plt.subplots(figsize = (10,10))
x_plot = np.random.normal(size=100)
y_plot = np.random.normal(size=100)

ax.scatter(x_plot, y_plot)
ax.set_xlabel("X")
ax.set_ylabel("Y")
ax.set_title("Scatter Plot of Random variables")
```

```python
fig2, ax2 = plt.subplots(nrows = 2, ncols = 2, figsize = (10,10))
ax2[0, 0].plot(x_plot, y_plot, 'o')
ax2[0, 1].plot(y_plot, x_plot, 'o')
ax2[1, 0].plot(x_plot, x_plot**2, 'o')
ax2[1, 1].plot(y_plot, y_plot**2, 'o')
```

Contour plots

```python
fig3, ax3 = plt.subplots(figsize = (10, 10))
x_contour = np.linspace(-np.pi, np.pi, 100)
y_contour = x_contour

f = np.multiply.outer(np.cos(x_contour), 1 / (1 + x_contour**2))
ax3.contourf(x_contour, y_contour, f, levels=500)
```

```python
fig4, ax4 = plt.subplots(figsize = (10, 10))
ax4.imshow(f)
```

```python
auto = pd.read_csv("Auto.csv", na_values = ['?'])
auto['horsepower'].sum()
```

```python
#Renaming the index to the names of the vehicles
auto_new = auto.set_index('name')
auto_new
```

```python
#Gettting all of the vehicles that have horsepower greater than 100
idx1 = auto_new['horsepower'] > 100
horse_auto = auto_new[idx1]
horse_auto
```

```python
#Getting the 1, 4 rows and 7, 8 columns
horse_auto.iloc[[0, 3], [6, 7]]
```

```python
#Getting the buick century, and the displacement and weight columns
horse_auto.loc['buick century', ['displacement', 'weight']]
```

```python
#Combining the boolean indexing and the loc method
idx2 = horse_auto['weight'] > 2000
horse_auto.loc[idx2, ['displacement', 'weight']]
```
